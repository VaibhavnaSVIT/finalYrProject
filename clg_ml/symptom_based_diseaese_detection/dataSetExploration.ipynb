```python
import pandas as pd

# Paths to the datasets
datasets = {
    'Training': 'Training.csv',
    'Testing': 'Testing.csv',
    'Dataset': 'dataset.csv'
}

for name, path in datasets.items():
    print(f"\n📁 Analyzing {name} Dataset ({path})")
    print("-" * 50)
    
    try:
        df = pd.read_csv(path)
    except Exception as e:
        print(f"❌ Failed to read {path}: {e}")
        continue

    print(f"✅ Loaded successfully: {df.shape[0]} rows × {df.shape[1]} columns\n")
    
    print("🔸 Column names:")
    print(df.columns.tolist())

    print("\n🔸 First 5 rows:")
    print(df.head())

    if 'prognosis' in df.columns:
        print(f"\n🔸 Unique diseases in 'prognosis': {df['prognosis'].nunique()}")
        print(df['prognosis'].value_counts())

    print("\n🔸 Missing values:")
    print(df.isnull().sum())
    
    print("=" * 60)
```
